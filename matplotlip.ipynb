```python
total_number=a['type'].value_counts()
plt.bar(total_number.index,total_number.values)
plt.show()
```

```python
content_percentage=a['type'].value_counts()
plt.pie(content_percentage.values,labels=content_percentage.index,autopct='%1.1f%%')
plt.show()
```

```python
movie_duration=a['duration'].value_counts()
plt.hist(movie_duration.values)
plt.show()
```

```python
countries = a['country'].value_counts().head(15)
plt.barh(countries.index, countries.values)
plt.xlabel('Count')
plt.ylabel('Countries')
plt.title('Top 15 countries')
plt.show()
```

```python
content_released=a['release_year'].value_counts()
plt.plot(content_released.index,content_released.values)
plt.show()
```

```python
movies = a[a['type'] == 'Movie']
tv = a[a['type'] == 'TV Show']

plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.hist(movies['release_year'],bins=20, edgecolor='black')
plt.xlabel('Release Year')
plt.ylabel('Count')

plt.subplot(1,2,2)
plt.hist(tv['release_year'],bins=20, edgecolor='black')
plt.xlabel('Release Year')
plt.ylabel('Count')
```

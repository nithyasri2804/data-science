```python
x=s[['show_id','cast','date_added','rating', 'duration', 'listed_in', 'description']]
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss=StandardScaler()
```

```python
x=ss.fit_transform(x)
x
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
model=NearestNeighbors(n_neighbors=6,metric="euclidean")
model.fit(x)
```

```python
def rec(song):
    result = s[s["title"].str.strip().str.lower() == song.strip().lower()]
    if result.empty:
        return "Movie/Song not found in dataset"
    s_index = result.index[0]
    query_point = s.loc[[s_index], x].values
    distances, indices = model.kneighbors(query_point)
    return s.iloc[indices[0]]["title"]
print(rec("Monster Maker"))
```
